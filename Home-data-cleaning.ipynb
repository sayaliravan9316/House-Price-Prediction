```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('clean.csv')
```

```python
df.info()
```

```python
df=df.drop(columns=['Facing','Society','Ownership','Basement','Main Road','Garden/Park','Pool','latitude','longitude'])
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Carpet Area']=df['Carpet Area'].fillna(df['Carpet Area'].mean())
```

```python
df['Area(in sqft)']=df['Area(in sqft)'].fillna(df['Area(in sqft)'].mean())
```

```python
df['FlatFloor']=df['FlatFloor'].fillna(df['FlatFloor'].mean())
```

```python
df['TotalFloors']=df['TotalFloors'].fillna(df['TotalFloors'].mean())
```

```python
df.isnull().sum()
```

```python
df=df.dropna(subset=['SalePrice(in Crores)'])
```

```python
df.info()
```

```python
df['Bathroom']=pd.to_numeric(df['Bathroom'],errors='coerce')
```

```python
df.dropna(subset=['Bathroom'],inplace=True)
```

```python
df['Balcony']=pd.to_numeric(df['Balcony'],errors='coerce')
```

```python
df.dropna(subset=['Balcony'],inplace=True)
```

```python
df['BHK']=pd.to_numeric(df['BHK'],errors='coerce')
```

```python
df.dropna(subset=['BHK'],inplace=True)
```

```python
df.duplicated().value_counts()
df.drop_duplicates(inplace=True)
```
